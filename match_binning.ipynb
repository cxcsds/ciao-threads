## create image

## what are image boundaries?

### dmlist subspace

```bash
dmlist img.fits subspace | grep sky
```

### dmcoords

```bash
punlearn dmcoords
dmcoords img.fits opt=logical logicalx=0.5 logicaly=0.5 verb=0
pget dmcoords x y
```

```bash
punlearn dmcoords
dmcoords img.fits opt=logical logicalx=294.5 logicaly=297.5 verb=0
pget dmcoords x y
```

## Caveats

```bash
dmcopy "acisf01838N003_evt2.fits[bin x=3595.5:4771.5:#294,y=3799.5:4987.5:#297]" img2.fits clob+
```

```bash
dmimgcalc img.fits img2.fits diff.fits sub clob+
```

```bash
dmstat diff.fits centroid=no
```

```bash
dmcopy "acisf01838N003_evt2.fits[bin x=3570:4874:16,y=3650:4980:16]" zoom16_1.fits cl+
dmcopy "acisf01838N003_evt2.fits[x=3570:4874,y=3650:4980][bin x=::16,y=::16]" zoom16_2.fits cl+
dmimgcalc zoom16_1.fits zoom16_2.fits zoom_diff.fits sub cl+
dmstat zoom_diff.fits centroid=no
```

```bash
ds9 zoom_diff.fits -scale linear \
  -cmap load $ASCDS_CONTRIB/data/004-phase.lut \
  -zoom to fit -scale limits -50 50 -saveimage png ds9.png -quit

display < ds9.png
```
