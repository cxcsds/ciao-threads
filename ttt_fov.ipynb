```bash
cat << EOM > p3.py
from region import *
mycelreg = regParse("region(acis_fov.fits[cols eqpos,shape,component])")
pointing=regInsideRegion(mycelreg, 151.1395864022, 41.2326566497 )
random=regInsideRegion(mycelreg, 145.2, 45.0 )
print("Is _PNT inside FOV? {}".format(pointing))
print("Is random location inside FOV? {}".format(random))
EOM

python p3.py
```
