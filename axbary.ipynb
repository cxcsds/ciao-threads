# axbary

```bash
source /export/ciao/bin/ciao.bash -o
PFILES="./param;$ASCDS_INSTALL/param:$ASCDS_INSTALL/contrib/param"
mkdir -p ./param
ASCDS_WORK_PATH=./
```

## Getting started

```bash
/bin/rm -rf 133
download_chandra_obsid 133 evt2,eph1,asol
mv -f 133/primary/*fits.gz .
gunzip -f *gz
```

```bash
ls -1 *fits
```

```bash
dmkeypar acisf00133N003_evt2.fits TSTART echo+
```

## Set RA and DEC

```bash
dmlist acisf00133N003_evt2.fits header | egrep '(RA|DEC)_TARG'
```

## Run Tool

```bash
punlearn axbary
pset axbary infile=acisf00133N003_evt2.fits
pset axbary orbitfile=orbitf051580864N002_eph1.fits
pset axbary outfile=acis_133_bary_evt2.fits 
pset axbary ra=85.04650 dec=-69.33164  
axbary mode=h clobber=yes
```

## Update asol

```bash
punlearn axbary
pset axbary infile=pcadf052020859N003_asol1.fits
pset axbary orbitfile=orbitf051580864N002_eph1.fits
pset axbary outfile=pcad_133_bary_asol1.fits 
pset axbary ra=85.04650 dec=-69.33164  
axbary mode=h clobber=yes
```

```bash
dmhedit acis_133_bary_evt2.fits file= op=add key=ASOLFILE value="pcad_133_bary_asol1.fits" 
```

## Check output

```bash
dmlist acisf00133N003_evt2.fits"[cols time]" data rows=1:5
```

```bash
dmlist acis_133_bary_evt2.fits"[cols time]" data rows=1:5
```
