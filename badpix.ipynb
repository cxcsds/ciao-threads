## acis_set_ardlib

```bash
punlearn ardlib
```

```bash
acis_set_ardlib 1843/primary/*bpix1.fits.gz
```

### Manually

```bash
dmkeypar 1843/primary/acisf01843N003_evt2.fits.gz DETNAM echo+
```

```bash
for d in 0 1 2 3 6 7
do
    pset ardlib \
    AXAF_ACIS${d}_BADPIX_FILE="1843/primary/acisf01843_000N003_bpix1.fits.gz[BADPIX${d}]"
done
```

```bash
pdump ardlib | grep AXAF_ACIS._BADPIX_FILE
```
