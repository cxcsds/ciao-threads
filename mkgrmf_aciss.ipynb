# mkgrmf_aciss


```bash
source /export/ciao/bin/ciao.bash -o
PFILES="./param;$ASCDS_INSTALL/param:$ASCDS_INSTALL/contrib/param"
mkdir -p ./param
ASCDS_WORK_PATH=./
```

## Download

```bash
 /bin/rm -rf 459
 download_chandra_obsid 459 
```

```bash
chandra_repro 459 out="./" clob+
```

## mkgrmf

```bash
pset mkgrmf order=1
pset mkgrmf grating_arm=HEG
pset mkgrmf outfile=heg_p1.rmf
pset mkgrmf obsfile="acisf00459_repro_evt2.fits[EVENTS]"
pset mkgrmf regionfile=acisf00459_repro_pha2.fits
pset mkgrmf detsubsys=ACIS-S3
pset mkgrmf wvgrid_arf=compute
pset mkgrmf wvgrid_chan=compute
mkgrmf mode=h clob+
```
