# Add grating spectra

aka `combine_grating_spectra`

```bash
#source /export/ciao/bin/ciao.bash -o
PFILES="./param;$ASCDS_INSTALL/param:$ASCDS_INSTALL/contrib/param"
mkdir -p ./param
ASCDS_WORK_PATH=./
```

```bash
/bin/rm -rf 8985 8986 
download_chandra_obsid 8985,8986 
```

```bash
chandra_repro 8985,8986 out=""
```

```bash
/bin/ls 898?/repro/*
```

## Run c_g_s


```bash
/bin/ls */repro/*pha2.fits > pha.lis
/bin/ls */repro/tg/*arf > arf.lis
/bin/ls */repro/tg/*rmf > rmf.lis
```

```bash
punlearn combine_grating_spectra

pset combine_grating_spectra infile=@pha.lis
pset combine_grating_spectra arf=@arf.lis
pset combine_grating_spectra rmf=@rmf.lis
pset combine_grating_spectra add_plusminus=yes
pset combine_grating_spectra outroot=h1743-322

combine_grating_spectra mode=h clob+
```
