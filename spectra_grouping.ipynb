```bash
download_chandra_obsid 459,460 pha2
mv -fv */primary/*pha2.fits.gz .
gunzip -f *pha2.fits.gz 
rmdir -p */primary
```

## dmtype2split

```bash
dmtype2split "acisf00459N004_pha2.fits[tg_part=1,tg_m=-3]" \
  "459_heg_m1.fits[SPECTRUM]" clob+
```

```bash
dmkeypar 459_heg_m1.fits TG_M echo-
pget dmkeypar value
```

```bash
dmkeypar 459_heg_m1.fits TG_PART echo+
```

## Multiple spectra

```bash
dmtype2split hrcf00460N005_pha2.fits "leg_m1.fits[SPECTRUM], leg_p1.fits[SPECTRUM]" clob+
```

```bash
cat << EOM > output.lis 
leg_m1_b.fits[SPECTRUM]
leg_p1_b.fits[SPECTRUM]
EOM
```

```bash
dmtype2split hrcf00460N005_pha2.fits @output.lis clob+
```

## Group spectra

```bash
dmgroup infile="leg_p1.fits[SPECTRUM]" \
      outfile=leg_p1_bin10.pha \
      binspec="1:16384:10" grouptypeval="" grouptype=BIN ycolumn="counts" \
      xcolumn="channel" clob+
```

```bash
dmlist leg_p1_bin10.pha cols
```
