# Customizing an ACIS Bad Pixel File

```bash
source /export/ciao/bin/ciao.bash -o
mkdir -p ./param
PFILES="./param;$ASCDS_INSTALL/param:$ASCDS_INSTALL/contrib/param"
ASCDS_WORK_PATH=./
```

# Getting started

```bash
/bin/rm -rf 333
download_chandra_obsid 333 evt2,bpix,pbk
mv 333/primary/* 333/secondary/* ./
gunzip *fits.gz
```

```bash
dmmakepar acisf00333N003_evt2.fits 333_obs.par cl+
```

## A. Events on cols adjacent to bad cols

```bash
acis_build_badpix \
         pbkfile=acisf055915346N003_pbk0.fits \
         berrfile=none \
         calibfile=acisf00333_000N003_bpix1.fits \
         usrfile=none \
         bitflag="00000000000000022221100020022212" \
         procbias=no \
         biasfile=none \
         evt0file=none \
         obsfile=333_obs.par \
         outfile=acisf00333_000N003_col_bpix1.fits \
         cl+
```

## B. Events on chip node boundaries

```bash
acis_build_badpix \
         pbkfile="acisf055915346N003_pbk0.fits" \
         berrfile="none" \
         calibfile="acisf00333_000N003_bpix1.fits" \
         usrfile="none" \
         bitflag="00000000000000022220100020022222" \
         procbias=no \
         biasfile="none" \
         evt0file="none" \
         obsfile="333_obs.par" \
         outfile="acisf00333_000N003_node_bpix1.fits" \
         cl+
```

## C.  Events w/ afterglows

```bash
acis_build_badpix \
         pbkfile="acisf055915346N003_pbk0.fits" \
         berrfile="none" \
         calibfile="acisf00333_000N003_bpix1.fits" \
         usrfile="none" \
         bitflag="00000000000000020221100020022222" \
         procbias=no \
         biasfile="none" \
         evt0file="none" \
         obsfile="333_obs.par" \
         outfile="acisf00333_000N003_aft_bpix1.fits" \
         cl+
```

## `usrfile` parameter

```bash
cat << EOM > acisf00333_000N003_usr_bpix1.txt 
#ccd_id chipx_lo chipx_hi chipy_lo chipy_hi time time_stop bit action
#------ -------- -------- -------- -------- ---- --------- --- ------
 4      330      330      1        1024     0    0         1   include
 7      647      647      749      749      0    0         0   include
 5      608      608      282      282      0    0         -1  exclude
 6      585      585      1        1024     0    0         -1  exclude
EOM
```

```bash
acis_build_badpix \
         pbkfile="acisf055915346N003_pbk0.fits" \
         berrfile="none" \
         calibfile="acisf00333_000N003_bpix1.fits" \
         usrfile="acisf00333_000N003_usr_bpix1.txt" \
         bitflag="00000000000000022221100010022222" \
         procbias=no \
         biasfile="none" \
         evt0file="none" \
         obsfile="333_obs.par" \
         outfile="acisf00333_000N003_usr_bpix1.fits" \
         cl+
       
```

## Reprocess


```bash
download_chandra_obsid 333
```

```bash
cd 333/primary
mv acisf00333_000N003_bpix1.fits.gz acisf00333_000N003_bpix1.ORIG
```

```bash
cp ../../acisf00333_000N003_usr_bpix1.fits ./
cd ../..
```

```bash
ASCDS_WORK_PATH=`pwd`
/bin/rm -rf 333/repro
chandra_repro 333 out='' badpixel=no clob+
```

```bash
dmkeypar 333/repro/acisf00333_repro_evt2.fits BPIXFILE echo+
```

# Cleanup

```bash
mv -fv 333/repro/acisf00333_repro_evt2.fits ./
/bin/rm -rf 333
/bin/rm -f acisf00333N003_evt2.fits acisf00333_000N003_bpix1.fits acisf055915346N003_pbk0.fits
```
