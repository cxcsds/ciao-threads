# DM cube

```bash
source /export/ciao/bin/ciao.bash -o\n
PFILES="./param;$ASCDS_INSTALL/param:$ASCDS_INSTALL/contrib/param"
mkdir -p ./param
ASCDS_WORK_PATH=./
```

```bash
/bin/rm -rf 1463
download_chandra_obsid 1463 evt2
gunzip 1463/primary/acisf01463N006_evt2.fits.gz
mv 1463/primary/acisf01463N006_evt2.fits .
```

## Create Cube

```bash
dmlist "acisf01463N006_evt2.fits[bin x=::8,y=::8,time=::1000]" cols
```

```bash
dmlist "acisf01463N006_evt2.fits[bin x=3900:4400,y=4100:4600,time=::1000]" cols
```

```bash
dmcopy "acisf01463N006_evt2.fits[bin x=3900:4400,y=4100:4600,time=::1000]" jupiter_cube.fits cl+
```

```bash
dmlist jupiter_cube.fits blocks
```

## Display Cube

```bash
ds9 jupiter_cube.fits -log -saveimage png ds9_01.png -quit
display < ds9_01.png
```

```bash
ds9 -3d jupiter_cube.fits -log -3d az 60 -3d el 30 -saveimage png ds9_02.png -quit
display < ds9_02.png
```

## Manipulate Cube

```bash
dmcopy "jupiter_cube.fits[time=59969680:59973180]" range_cube.fits clob+
```

```bash
dmlist range_cube.fits cols
```

## Region Filtering

```bash
cat << EOM > circle.reg 
# Region file format: CIAO version 1.0
circle(4143.5,4266.5,137.16489)
EOM
```

```bash
ds9 jupiter_cube.fits -log -region circle.reg -saveimage png ds9_03.png -quit
display < ds9_03.png
```

```bash
dmcopy "jupiter_cube.fits[sky=region(circle.reg)][opt null=-99]" region_cube.fits cl+
```

```bash
dmlist region_cube.fits cols
```

```bash
ds9 region_cube.fits -log -region circle.reg -saveimage png ds9_04.png -quit
display < ds9_04.png
```

## Slicing

```bash
dmcopy "jupiter_cube.fits[#3=10][bin x,y]" plane10.fits cl+
```

```bash
dmlist plane10.fits cols
```

```bash
ds9 plane10.fits -log -saveimage png ds9_05.png -quit
display < ds9_05.png
```
