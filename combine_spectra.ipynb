# Combine spectra

```bash
source /export/ciao/bin/ciao.bash -o
PFILES="./param;$ASCDS_INSTALL/param:$ASCDS_INSTALL/contrib/param"
mkdir -p ./param
ASCDS_WORK_PATH=./
```

## Get Started

```bash
/bin/rm -rf 10289 10290 10291 10292 10293
download_chandra_obsid 10289,10290,10291,10292,10293 
```

```bash
cat << EOM > ds9.reg
# Region file format: DS9 version 4.1
global color=green dashlist=8 3 width=1 font="helvetica 10 normal roman" select=1 highlite=1 dash=0 fixed=0 edit=1 move=1 delete=1 include=1 source=1
fk5
circle(17:16:59.592,-62:49:13.868,4.976")
EOM


cat << EOM > bkg.reg
# Region file format: DS9 version 4.1
global color=green dashlist=8 3 width=1 font="helvetica 10 normal roman" select=1 highlite=1 dash=0 fixed=0 edit=1 move=1 delete=1 include=1 source=1
fk5
annulus(17:16:59.592,-62:49:13.868,5",9")
EOM

```

```bash
for d in 10289 10290 10291 10292 10293
do
  specextract ${d}/primary/*evt2.fits.gz"[sky=region(ds9.reg)]" \
  outroot=${d} bkgfile=${d}/primary/*evt2.fits.gz"[sky=region(bkg.reg)]" \
  mode=h clob+
done
```

```bash
/bin/ls -1 *.pi
```

```bash
/bin/ls -1 *.arf
```

```bash
/bin/ls -1 *.rmf
```

## Running combine_spectra

```bash
punlearn combine_spectra
pset combine_spectra src_spectra="10289.pi,10290.pi,10291.pi,10292.pi,10293.pi"
pset combine_spectra outroot=ngc6300
combine_spectra mode=h clob+
```

## Manually input stacks

```bash
/bin/ls ?????.pi > src.lis
/bin/ls ?????.arf > arf.lis
/bin/ls ?????.rmf > rmf.lis

# Repeat for background files
/bin/ls ?????_bkg.pi > bg.lis
/bin/ls ?????_bkg.arf > bg_arf.lis
/bin/ls ?????_bkg.rmf > bg_rmf.lis
```

```bash
cat src.lis
```

```bash
cat bg_arf.lis
```

```bash
punlearn combine_spectra
pset combine_spectra src_spectra=@src.lis
pset combine_spectra src_arfs=@arf.lis
pset combine_spectra src_rmfs=@rmf.lis
pset combine_spectra bkg_spectra=@bg.lis
pset combine_spectra bkg_arfs=@bg_arf.lis
pset combine_spectra bkg_rmfs=@bg_rmf.lis
pset combine_spectra outroot=ngc6300_stacks
combine_spectra mode=h clob+
```
