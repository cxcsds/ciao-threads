# HRC badpix

```bash
source /export/ciao/bin/ciao.bash -o
PFILES="./param;$ASCDS_INSTALL/param:$ASCDS_INSTALL/contrib/param"
mkdir -p ./param
ASCDS_WORK_PATH=./
```

```bash
download_chandra_obsid 6476 evt1,bpix
mv -f 6476/secondary/*.fits.gz .
gunzip -f *fits.gz
rmdir -p 6476/secondary
```

## Create obs.par

```bash
dmmakepar hrcf06476_000N004_evt1.fits hrcf06476_obs.par clob+
```

```bash
plist hrcf06476_obs.par 
```

```bash
pget hrcf06476_obs.par detnam tstart tstop
```

## Update bpix file to current CALDB

```bash
pset hrc_build_badpix infile=CALDB
pset hrc_build_badpix degapfile=CALDB
pset hrc_build_badpix outfile=hrcf06476_bpix1_caldb.fits
pset hrc_build_badpix obsfile=hrcf06476_obs.par 
hrc_build_badpix mode=h clob+
```

## Check degap file

```bash
dmkeypar hrcf06476_000N004_evt1.fits DEGAP echo+
```

```bash
dmkeypar hrcf06476_000N004_bpix1.fits DEGAP echo+
```

```bash
dmkeypar hrcf06476_bpix1_caldb.fits DEGAP echo+
```

## Specify degap

```bash
pset hrc_build_badpix infile=CALDB
pset hrc_build_badpix degapfile=$CALDB/data/chandra/hrc/gaplookup/hrciD1999-07-22gaplookupN0003.fits
pset hrc_build_badpix outfile=hrcf06476_bpix1_match.fits
pset hrc_build_badpix obsfile=hrcf06476_obs.par 
hrc_build_badpix mode=h clob+ 
```

# Cleanup


```bash
/bin/rm -f hrcf06476_000N004_evt1.fits hrcf06476_000N004_bpix1.fits
```
