### Determine chips

```bash
dmstat "acisf00953N004_evt2.fits[sky=region(src1.reg)][cols ccd_id]"
```

```bash
dmstat "acisf00953N004_evt2.fits[sky=region(src2.reg)][cols ccd_id]"
```

```bash
dmstat "acisf00953N004_evt2.fits[sky=region(bkg.reg)][cols ccd_id]"
```

### background subtracted lightcurve

```bash
pset dmextract infile="acisf00953N004_evt2.fits[ccd_id=3,sky=region(src2.reg)][bin time=::2000]"
pset dmextract outfile=src2_sub_lc.fits
pset dmextract bkg="acisf00953N004_evt2.fits[ccd_id=3,sky=region(bkg.reg)]"
pset dmextract opt=ltc1
dmextract mode=h clob+
```

```bash
dmlist src2_sub_lc.fits cols
```

```bash
cat << EOM > c1.py
make_figure("src2_sub_lc.fits[cols dt,net_rate,err_rate]", "line.color=red")
print_window("chips_01.png","export.clobber=True")
quit()
EOM

chips -b c1.py

display < chips_01.png
```

```bash
dmlist "src2_sub_lc.fits[cols time,dt,count_rate]" data
```
