### dmextract


```bash
dmextract "acisf01843N003_evt2.fits[bin pha=1:2048:2]" histogram.pha cl+
```

```bash
cat << EOM > ds9.reg
# Region file format: CIAO version 1.0
ellipse(1628,4116,92,172,0)
EOM

dmextract "acisf01843N003_evt2.fits[sky=region(ds9.reg)][bin pi]" \
      spectrum.fits cl+
```

```bash
cat << EOM > infiles.lis
acisf01843N003_evt2.fits[bin pha=1:2048:2]
acisf01843N003_evt2.fits[sky=ellipse(1628,4116,92,172,0)][bin pi]
EOM

cat << EOM >  outfiles.lis
histogram_lis.pha
spectrum_lis.fits
EOM

dmextract infile=@infiles.lis outfile=@outfiles.lis cl+
```

### dmgti

```bash
dmextract \
      "acisf01843N003_evt2.fits[(x,y)=field()-ellipse(1628,4116,100,140,0)][bin time=::3.24104]" \
      bkg_lc.fits opt=ltc1 cl+
```

```bash
dmlist "bkg_lc.fits[cols time,count_rate,stat_err][#row=1:3,490:495,740:745,3000:3005]" data
```

```bash
dmsort bkg_lc.fits sorted_bkg_lc.fits keys=-count_rate cl+
dmlist "sorted_bkg_lc.fits[cols time,count_rate,stat_err]" data row=1:10
```

```bash
dmgti infile=bkg_lc.fits outfile=acis_gti.fits \
      userlimit="(count_rate>40.0)&&(count_rate<100.0)" clobber=yes
```

```bash
dmlist "acis_gti.fits[gti]" data
```

```bash
dmcopy "acisf01843N003_evt2.fits[@acis_gti.fits]" acis_filtered_evt2.fits cl+
```
