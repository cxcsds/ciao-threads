## Running multiple instances

```bash
apply_fov_limits acisf01843N004_evt2.fits[energy=500:7000] img.fits clob+ bin=8
```

```bash
paccess aconvolve
```

```bash
cp -fv `paccess aconvolve` ./aconvolve.g3.par
chmod +w ./aconvolve.g3.par
```

```bash
cp -fv `paccess aconvolve` ./aconvolve.g5.par
chmod +w ./aconvolve.g5.par
```

```bash
pset aconvolve.g3.par infile=img.fits method=fft cl+
pset aconvolve.g3.par outfile=img.sm3.fits 
pset aconvolve.g3.par kernelspec="lib:gauss(2,5,1,3,3)"
```

```bash
pset aconvolve.g5.par infile=img.fits method=fft cl+
pset aconvolve.g5.par outfile=img.sm5.fits 
pset aconvolve.g5.par kernelspec="lib:gauss(2,5,1,5,5)"
```

```bash
aconvolve @@aconvolve.g3 mode=h && aconvolve @@aconvolve.g5 mode=h
```
